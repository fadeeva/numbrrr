# Linear Congruential Generator (LCG)

```python
import numpy as np

import matplotlib.pyplot as plt

jtplot.style(figsize=(24, 5))
```

```python
def LCG(multiplier_a, increment_c, seed, module_m):
    return (multiplier_a * seed + increment_c) % module_m
```

```python
N = 1_000

seed = 1
a = 2
c = 0
m = 9

X_n = []
for i in range(N):
    X_n.append(seed)
    seed = LCG(a, c, seed, m)
    
```

```python
def lcg(modulus: int, a: int, c: int, seed: int):
    """Linear congruential generator from Wikipedia."""
    while True:
        seed = (a * seed + c) % modulus
        yield seed
```

```python
seed_wiki = 1
for _, ran in zip(range(10), lcg(m, a, c, seed_wiki)):
    print(ran, end=' ')
```
