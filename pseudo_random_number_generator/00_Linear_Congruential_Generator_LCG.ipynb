# Linear Congruential Generator (LCG)

$$
X_{n+1} = (aX_n + c)\mod m
$$

```python
def LCG(multiplier_a, increment_c, seed, module_m):
    return (multiplier_a * seed + increment_c) % module_m
```

```python
N = 1_000

seed = 1
a = 2
c = 0
m = 9

def lcg_sequence(seed, a, c, m, n):
    for _ in range(n):
        yield seed
        seed = LCG(a, c, seed, m)

X_n = list(lcg_sequence(seed, a, c, m, N))
```

```python
plt.figure(figsize=(18, 3))

plt.scatter(np.arange(100), X_n[:100])
plt.title('LCG first 100 numbers')
```

```python
def lcg(modulus: int, a: int, c: int, seed: int):
    """Linear congruential generator from Wikipedia."""
    while True:
        seed = (a * seed + c) % modulus
        yield seed
```

```python
seed_wiki = 1
for _, ran in zip(range(10), lcg(m, a, c, seed_wiki)):
    print(ran, end=' ')
```
